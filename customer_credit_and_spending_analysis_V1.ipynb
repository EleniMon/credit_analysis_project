# Customer Credit & Spending Analysis

### Importing Relevant Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Dataset Loading

```python
raw_df = pd.read_csv('customer_credit_and_spending.csv')
```

## Initial Data Exploration

```python
# Displaying raw df information
raw_df.info()
raw_df.head(10)
```

## Data Cleaning

```python
# Creating a separate copy for cleaning
clean_df = raw_df.copy()

# Removing columns not necessary for the analysis
clean_df = clean_df.drop(columns=["phone_number", "email", "zip_code", "state", "last_login", "account_created", "last_payment_date"])

# Checking if any customer ID appears more than once
print("Duplicate customer IDs:", clean_df["customer_id"].duplicated().sum())
```

```python
# Removing customer ID column
clean_df = clean_df.drop("customer_id", axis=1)
```

```python
# Inspecting the unique values in each column
for col in clean_df.columns:
    print(col)
    print(clean_df[col].unique())
    print()
```

### Data Formatting

```python
# Correcting formatting and spelling issues
clean_df["education"] = clean_df["education"].replace({
    "HighSchool": "High School"
})
clean_df["marital_status"] = clean_df["marital_status"].replace({
    "Divorsed": "Divorced"
})
clean_df["income"] = clean_df["income"].replace({
    "35k": "35000"
})
```

### Data Type Conversion

```python
# Columns containing numbers currently stored as text
numeric_columns = [
    "income",
    "credit_score",
    "employment_years",
    "default_status"
]
# Converting values to numbers and replacing invalid text with NaN
for col in numeric_columns:
    clean_df[col] = pd.to_numeric(clean_df[col], errors="coerce")
```

### Missing Values

```python
# Replacing negative ages with missing values
clean_df.loc[clean_df["age"] < 0, "age"] = float("nan")

# Replacing negative spending with missing values
clean_df.loc[clean_df["avg_monthly_spend"] < 0, "avg_monthly_spend"] = float("nan")

# Replacing unidentified product categories with missing values
clean_df["product_type"] = clean_df["product_type"].replace(
    ["Unknown", "3"], float("nan")
)
# Replacing the unclear purchase frequency with a missing value
clean_df["purchase_frequency"] = clean_df["purchase_frequency"].replace(
    "30", float("nan")
)
```

```python
# Displaying number of NaNs per column
print(clean_df.isna().sum())
```

`Missing values were kept as NaN because the correct values are unknown.
Filling them with an average or median would add guesses that could affect the findings.
Rows were not deleted because the dataset only has 50 customers, and those rows still contain useful information in other columns.`
